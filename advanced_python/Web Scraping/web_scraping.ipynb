## 1. Import Required Libraries

First, we import all the necessary libraries for web scraping and data handling.

```python
# Web Scraping Remote Job Listings

# This notebook demonstrates how to scrape remote job listings from the Remotive API, process the data, and save it to CSV files. It is intended for educational purposes and covers basic web scraping, data extraction, and file handling in Python.

# ## Libraries Used
# - `requests`: For making HTTP requests to the API
# - `BeautifulSoup`: For parsing HTML content in job descriptions
# - `csv`: For writing data to CSV files
# - `time`: For adding delays between requests to avoid overloading the server

import requests
from bs4 import BeautifulSoup
import csv
import time
```

## 2. Scrape Multiple Job Categories

Now, let's extend our script to fetch jobs from several categories and save all results to a single CSV file. We'll add a short delay between requests to be polite to the API.

```python
# Fetch and Save Software Development Jobs
response = requests.get("https://remotive.com/api/remote-jobs?category=software-dev")
jobs = response.json().get("jobs", [])

# Save to CSV
with open("software_jobs.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Title", "Company", "Location", "Job Type", "Salary", "URL", "Description"])

    for job in jobs:
        desc = BeautifulSoup(job.get("description", ""), "html.parser").get_text().strip()
        writer.writerow([
            job.get("title", "N/A"),
            job.get("company", "N/A"),
            job.get("candidate_required_location", "N/A"),
            job.get("job_type", "N/A"),
            job.get("salary", "N/A"),
            job.get("url", "N/A"),
            desc[:300]
        ])

print(f"Done! Saved {len(jobs)} jobs to software_jobs.csv")
```

```python
categories = [
    "software-dev", "marketing", "hr", "devops", "data"
]

with open("all_jobs.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Category", "Title", "Company", "Location", "Job Type", "Salary", "URL", "Description"])

    for cat in categories:
        print(f"Scraping category: {cat}")
        url = f"https://remotive.com/api/remote-jobs?category={cat}"
        jobs = requests.get(url).json().get("jobs", [])

        for job in jobs:
            desc = BeautifulSoup(job.get("description", ""), "html.parser").get_text().strip()
            writer.writerow([
                cat,
                job.get("title", "N/A"),
                job.get("company", "N/A"),
                job.get("candidate_required_location", "N/A"),
                job.get("job_type", "N/A"),
                job.get("salary", "N/A"),
                job.get("url", "N/A"),
                desc[:300]
            ])
        time.sleep(2)

print("Done! Saved jobs to all_jobs.csv")
```
