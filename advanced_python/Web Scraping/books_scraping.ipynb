# Web Scraping Books: Books to Scrape

This notebook demonstrates how to scrape book data from the 'Books to Scrape' website and save it to a CSV file.

```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

## Fetch and Parse Webpage
Fetch the webpage using requests and parse the HTML content with BeautifulSoup.

```python
# Fetch the webpage
response = requests.get("https://books.toscrape.com/index.html").text

# Parse the HTML content
soup = BeautifulSoup(response, "lxml")
```

## Extract Book Information
Extract book titles, prices, stock availability, and star ratings from the parsed HTML.

```python
# Initialize lists to store book data
titles = []
prices = []
in_stocks = []
stars = []

# Extract book information
books = soup.find_all("li", class_="col-xs-6 col-sm-4 col-md-3 col-lg-3")

for book in books:
    book_title = book.find("h3").text
    book_price = book.find("p", class_="price_color").text
    book_in_stock = book.find("p", class_="instock availability").text.strip()
    star_p = book.find("p", class_="star-rating")
    rating_word = star_p["class"][1]  # Extract the second class which is the rating

    titles.append(book_title)
    prices.append(book_price)
    in_stocks.append(book_in_stock)
    stars.append(rating_word)
```

## Create and Display DataFrame
Create a pandas DataFrame with the extracted data and display it.

```python
book_df = pd.DataFrame({
    "Title": titles,
    "Price": prices,
    "In Stock": in_stocks,
    "Star Rating": stars
})

book_df
```

## Save DataFrame to CSV
Save the DataFrame to a CSV file named 'books_scraped.csv'.

```python
book_df.to_csv("books_scraped.csv", index=False)
print("Data saved to books_scraped.csv")
```
